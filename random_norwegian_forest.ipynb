```python
# Correlation between features
df_train_corr = df_train.corr()

# Plot
plt.figure(figsize=(8, 6))
sns.heatmap(data=df_train_corr,
            annot=True,
            annot_kws={"size":7},
            
            fmt=".2f"
            )
plt.xticks(rotation=45,
           ha='right')

# Save the figure
plt.savefig('feature_heatmap.png')
```
